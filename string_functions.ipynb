```python
# split an email string to username , domain and sub domain
```

```python
email = 'email@gmail.com'
username, domain = email.split('@')
subdomain, domain = domain.split('.')
print(username)
print(domain)
print(subdomain)
```

```python
# Now find the index position of @ and . in email string
dot_pos = email.find('.')
at_pos = email.find('@')
print(dot_pos)
print(at_pos)
#
```
