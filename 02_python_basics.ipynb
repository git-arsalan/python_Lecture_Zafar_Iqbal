**String can be defined in single quotes or double quotes**

```python
"Hello python" # double quotes
#'Hello python' # single quotes
```

```python
"printing string with 'single quotes'" # we cannnot print string that contains single quotes inside and defined using single quotes
```

```python
'printing with "Double Quotes"' # we cannot print string that contains double quotes inside and  defined using double quotes
```

```python
# using escape sequence character in string
print('printing with \'single quotes\'')
# What if we want to print a escape sequence char?
print('printing with \\single quotes\\')
#OR
print('printing with \\\\single quotes\\\\')
```

```python
# escape sequence printing using raw format
print(r"path to windows: C:\windows\\") # Added an extra backslash to properly terminate the raw string
#
```
