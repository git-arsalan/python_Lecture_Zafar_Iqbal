```python
# Print table of any number using for loop in python
```

```python
table_Num = int(input('Please Enter Number to Print its Table'))
for n in range(1 , 11):
  print(f"{table_Num} X  {n} = {table_Num * n}")

```

```python
# Now Write a Program to calculate Factorial of a number
```

```python
input_Num = int(input('Please Enter Number to Print its Factorial'))
fact_Num = input_Num
fact = 1
for n in range(1 , input_Num + 1):
  fact = fact * n
print(f"The factorial of {input_Num} is : {fact}")
```

```python
# by using While loop
input_Num = int(input('Please Enter Number to Print its Factorial'))
fact_Num = input_Num
fact = 1
while fact_Num != 1:
  fact = fact * fact_Num
  fact_Num = fact_Num - 1
print(f"The factorial of {input_Num} is : {fact}")
```
