```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn import linear_model
```

```python
# Importing Data
df = pd.read_excel('/content/drive/MyDrive/ML_DATA/SLR_Data.xlsx')
df
```

```python
plt.xlabel('MILEAGE' , color='red')
plt.ylabel('PRICE', color='blue')
plt.title('MILEAGE VS PRICE')
plt.scatter(df.MILEAGE,df.PRICE)
```

```python
# Now Applying Linear Model Regression to our Dataset
lr_Model = linear_model.LinearRegression()
lr_Model.fit(df[['MILEAGE']], df['PRICE']) # used to train the Model on basis of provided data

```

```python
# Now Predicting Unknown mileage from our trained model
lr_Model.predict([[50000]])
```

```python
# Now calculating the score for our model
lr_Model.score(df[['MILEAGE']], df['PRICE'])
```

```python
# Finding the coefficent and intercept for the trained model
# y = wx + b
print(f"Model coefficent {lr_Model.coef_}")
print(f"Model Intercept {lr_Model.intercept_}" )
```
