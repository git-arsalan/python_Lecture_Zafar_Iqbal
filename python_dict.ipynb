```python
# Dictionary is mutable(changeable) datatype exists in key-value pair form
```

```python
employees = {1:"Hamid",
             2:"samson",
             3:"arjun",
             4:"David"}
print(employees)
type(employees)
```

```python
# to Get All the keys in Dict
employees.keys()
#type(employees.keys())
```

```python
print(type(employees.keys()))
print(list(employees.keys()))
print(type(list(employees.keys())))
```

```python
# to get all the values in Dict
employees.values()
# dict values are also list and can be converted into list datatype
print(type(employees.values()))
print(list(employees.values()))
#
```
