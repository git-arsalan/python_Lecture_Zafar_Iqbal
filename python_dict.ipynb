```python
# Dictionary is mutable(changeable) datatype exists in key-value pair form
```

```python
employees = {1:"Hamid",
             2:"samson",
             3:"arjun",
             4:"David"}
print(employees)
type(employees)
```

```python
# to Get All the keys in Dict
employees.keys()
#type(employees.keys())
```

```python
print(type(employees.keys()))
print(list(employees.keys()))
print(type(list(employees.keys())))
```

```python
# to get all the values in Dict
employees.values()
# dict values are also list and can be converted into list datatype
print(type(employees.values()))
print(list(employees.values()))
#
```

```python
# to get specific value against key in dict
employees[1]

# to set and retreive a message if key is not found
print(employees.get(1,"Key not found"))
print(employees.get(5,"Key not found"))
```

```python
# to pop an Item from dictionary
print(employees.pop(6, 'Key not found'))
print(employees.pop(1,'Key not found'))
print(employees)
```
