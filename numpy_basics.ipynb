```python
import numpy as np
np_array = np.array([[[1,2,3],[1,2,3],[4,5,6]],
                    [[1,2,3],[1,2,3],[4,5,6]],
                     [[1,2,3],[4,5,6],[4,5,6]]
                    ])
np_array
```

```python
print(f'dimension of np array is {np_array.ndim}')
print(f'shape of np array is {np_array.shape}')
# size of np array
print(f'size of np array is {np_array.size}')
```

```python
# np array doesnt contains homogenous shape. example
test_array = np.array([[1,2,3],[4,5,6]],
                       [[1,2],[1,2]])
```

```python
test_array = np.array([[[1,2,3],[4,5,6]],
                       [[1,2,3],[1,2,3]],
                       [[1,2,3],[1,2,3]]])
print(test_array.ndim)
print(test_array.size)
print(test_array.shape)
```

```python
test_array
```

```python
# chain indexing in numpy
print(test_array[0][1][2])
print(test_array[1][1][2])
```

```python
# multi-dimension indexing
print(test_array[3,0,1])
```
